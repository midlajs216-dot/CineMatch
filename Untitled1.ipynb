```python
import pandas as pd

# 1. Create DataFrame
data = {
    'Order_ID': range(1001, 1021),
        'Salesperson': [
                'Anu', 'Rahul', 'Meera', 'Arjun', 'Anu',
                        'Priya', 'Rahul', 'Meera', 'Arjun', 'Priya',
                                'Anu', 'Rahul', 'Meera', 'Arjun', 'Priya',
                                        'Anu', 'Rahul', 'Meera', 'Arjun', 'Priya'
                                            ],
                                                'Region': [
                                                        'North', 'South', 'East', 'West', 'North',
                                                                'South', 'East', 'West', 'North', 'South',
                                                                        'East', 'West', 'North', 'South', 'East',
                                                                                'West', 'North', 'South', 'East', 'West'
                                                                                    ],
                                                                                        'Category': [
                                                                                                'Electronics', 'Furniture', 'Clothing', 'Electronics',
                                                                                                        'Furniture', 'Clothing', 'Electronics', 'Furniture',
                                                                                                                'Clothing', 'Electronics', 'Furniture', 'Clothing',
                                                                                                                        'Electronics', 'Furniture', 'Clothing', 'Electronics',
                                                                                                                                'Furniture', 'Clothing', 'Electronics', 'Furniture'
                                                                                                                                    ],
                                                                                                                                        'Product': [
                                                                                                                                                'Laptop', 'Chair', 'Shirt', 'Mobile', 'Table',
                                                                                                                                                        'Jeans', 'Headphones', 'Desk', 'Jacket', 'Tablet',
                                                                                                                                                                'Sofa', 'Shoes', 'Monitor', 'Cabinet', 'Dress',
                                                                                                                                                                        'Printer', 'Bed', 'T-shirt', 'Camera', 'Bookshelf'
                                                                                                                                                                            ],
                                                                                                                                                                                'Quantity': [
                                                                                                                                                                                        5, 12, 30, 8, 6, 25, 15, 4, 10, 7,
                                                                                                                                                                                                2, 18, 6, 5, 12, 9, 3, 20, 4, 6
                                                                                                                                                                                                    ],
                                                                                                                                                                                                        'Unit_Price': [
                                                                                                                                                                                                                55000, 4500, 1200, 18000, 12000,
                                                                                                                                                                                                                        1800, 2500, 15000, 2200, 28000,
                                                                                                                                                                                                                                35000, 3000, 22000, 8000, 2500,
                                                                                                                                                                                                                                        6500, 40000, 900, 32000, 11000
                                                                                                                                                                                                                                            ]
                                                                                                                                                                                                                                            }

                                                                                                                                                                                                                                            df = pd.DataFrame(data)

                                                                                                                                                                                                                                            # 2. Calculate Total Sales
                                                                                                                                                                                                                                            df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

                                                                                                                                                                                                                                            # 3. First and last 5 records
                                                                                                                                                                                                                                            print("First 5 Records:")
                                                                                                                                                                                                                                            print(df.head())
                                                                                                                                                                                                                                            print("\nLast 5 Records:")
                                                                                                                                                                                                                                            print(df.tail())

                                                                                                                                                                                                                                            # 4. Dataset information and statistics
                                                                                                                                                                                                                                            print("\nDataset Information:")
                                                                                                                                                                                                                                            df.info()
                                                                                                                                                                                                                                            print("\nSummary Statistics:")
                                                                                                                                                                                                                                            print(df.describe())

                                                                                                                                                                                                                                            # 5. Region-wise total sales
                                                                                                                                                                                                                                            print("\nRegion-wise Total Sales:")
                                                                                                                                                                                                                                            print(df.groupby('Region')['Total_Sales'].sum())

                                                                                                                                                                                                                                            # 6. Category-wise average sales
                                                                                                                                                                                                                                            print("\nCategory-wise Average Sales:")
                                                                                                                                                                                                                                            print(df.groupby('Category')['Total_Sales'].mean())

                                                                                                                                                                                                                                            # 7. Top salesperson
                                                                                                                                                                                                                                            print("\nTop Salesperson:")
                                                                                                                                                                                                                                            print(df.groupby('Salesperson')['Total_Sales'].sum().idxmax())

                                                                                                                                                                                                                                            # 8. Sales above ₹50,000
                                                                                                                                                                                                                                            print("\nProducts with Sales Above ₹50,000:")
                                                                                                                                                                                                                                            print(df[df['Total_Sales'] > 50000])

                                                                                                                                                                                                                                            # 9. Top 5 by sales
                                                                                                                                                                                                                                            print("\nTop 5 Products:")
                                                                                                                                                                                                                                            print(df.sort_values(by='Total_Sales', ascending=False)[['Product', 'Total_Sales']].head(5))

                                                                                                                                                                                                                                            # 10. Region and Category-wise grouping
                                                                                                                                                                                                                                            print("\nRegion and Category-wise Analysis:")
                                                                                                                                                                                                                                            print(df.groupby(['Region', 'Category']).agg(
                                                                                                                                                                                                                                                Total_Sales=('Total_Sales', 'sum'),
                                                                                                                                                                                                                                                    Average_Sales=('Total_Sales', 'mean'),
                                                                                                                                                                                                                                                        Total_Quantity=('Quantity', 'sum')
                                                                                                                                                                                                                                                        ))

                                                                                                                                                                                                                                                        # 11. Missing values
                                                                                                                                                                                                                                                        print("\nMissing Values:")
                                                                                                                                                                                                                                                        print(df.isnull().sum())

                                                                                                                                                                                                                                                        # Handle missing values and remove duplicates
                                                                                                                                                                                                                                                        df = df.drop_duplicates()
                                                                                                                                                                                                                                                        df = df.fillna({
                                                                                                                                                                                                                                                            'Salesperson': 'Unknown',
                                                                                                                                                                                                                                                                'Region': 'Unknown',
                                                                                                                                                                                                                                                                    'Category': 'Unknown',
                                                                                                                                                                                                                                                                        'Product': 'Unknown',
                                                                                                                                                                                                                                                                            'Quantity': 0,
                                                                                                                                                                                                                                                                                'Unit_Price': 0,
                                                                                                                                                                                                                                                                                    'Total_Sales': 0
                                                                                                                                                                                                                                                                                    })

                                                                                                                                                                                                                                                                                    # Recalculate total sales
                                                                                                                                                                                                                                                                                    df['Total_Sales'] = df['Quantity'] * df['Unit_Price']

                                                                                                                                                                                                                                                                                    # 12. Export cleaned DataFrame to CSV
                                                                                                                                                                                                                                                                                    df.to_csv('cleaned_sales_data.csv', index=False)
                                                                                                                                                                                                                                                                                    print("\nCleaned data exported successfully!")

                                                                                                                                                                                                                                                                                    # Colab: download the CSV to your device
                                                                                                                                                                                                                                                                                    from google.colab import files
                                                                                                                                                                                                                                                                                    files.download('cleaned_sales_data.csv')
```
